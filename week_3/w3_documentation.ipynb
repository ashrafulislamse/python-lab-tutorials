# Week 3 Tutorial - Student Average Calculator

Documentation for `Student_Ave_Calc.py`. The program reads three quiz marks, calculates the average, decides pass or fail, and repeats for the next student.

## 1. Identify the Components

**Inputs**
- Quiz 1 mark (number)
- Quiz 2 mark (number)
- Quiz 3 mark (number)
- Choice to continue (Y/N)

**Process**
- Add the three marks and divide by 3 to get the average
- Compare the average with 50 using selection
- Repeat the whole block while the choice is Y (iteration)

**Output**
- The average mark
- "PASS" when the average is 50 or more, otherwise "FAIL"
- "Program Ended" when the loop stops

## 2. Design the Algorithm

### Pseudocode

```
START
    choice = "y"
    WHILE choice is "y" OR "Y"
        READ quiz_1, quiz_2, quiz_3
        average = (quiz_1 + quiz_2 + quiz_3) / 3
        PRINT average
        IF average >= 50 THEN
            PRINT "PASS"
        ELSE
            PRINT "FAIL"
        END IF
        READ choice
    END WHILE
    PRINT "Program Ended"
END
```

### Program structures used

| Structure | Where it appears in the program |
|---|---|
| Sequence | reading the three marks, then calculating and printing the average |
| Selection | `if average >= 50` decides PASS or FAIL |
| Iteration | `while` loop repeats for the next student until the choice is not Y |

````python
# Quick check of the calculation logic (the interactive version is in Student_Ave_Calc.py)
quiz_1 = 60.0
quiz_2 = 70.0
quiz_3 = 80.0
average = (quiz_1 + quiz_2 + quiz_3) / 3
print(f"Average mark: {average:.2f}")
if average >= 50:
    print("Result: PASS")
else:
    print("Result: FAIL")
````
